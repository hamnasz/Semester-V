```python
import nltk
import re
import string
import spacy
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.probability import FreqDist
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = """Prof. Johnson completed his PhD at Stanford University in 2015. 
He now works at Google Research developing machine learning models. 
Contact him at prof.johnson@gmail.com or visit https://research.google.com for more info!"""
```

```python
print("----- TASK 1: Display and Analyze the Text -----")
print("Original Text:\n", text)
```

```python
print("\n----- TASK 2: Word Tokenization using NLTK -----")

tokens = word_tokenize(text)
print("Tokens:", tokens)
print("\nTotal Tokens:", len(tokens))
print("Unique Tokens:", len(set(tokens)))

# Frequency distribution
freq = FreqDist(tokens)
print("\nToken Frequencies:")
for word, count in freq.items():
    print(f"{word}: {count}")
```

```python
print("\n----- TASK 3: Sentence Tokenization using NLTK -----")

sentences = sent_tokenize(text)
print("Sentences:", sentences)
print("Total Sentences:", len(sentences))

# Longest and shortest sentence by length
longest = max(sentences, key=len)
shortest = min(sentences, key=len)
print("\nLongest Sentence:", longest)
print("Shortest Sentence:", shortest)
```

```python
print("\n----- TASK 4: Tokenization using Regular Expressions -----")

# Remove punctuation and split on whitespace
regex_tokens = re.findall(r'\b\w+\b', text)
print("Regex Tokens:", regex_tokens)
print("\nTotal Regex Tokens:", len(regex_tokens))

# Comparison
print("\nComparison:")
print("NLTK Tokens count:", len(tokens))
print("Regex Tokens count:", len(regex_tokens))
print("Observation: Regex removed punctuation completely, while NLTK preserved symbols like '.' and '-'.")
print("Hence, NLTK handled punctuation contextually better.")
```

```python
print("\n----- TASK 5: Tokenization using SpaCy -----")

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

spacy_tokens = [token.text for token in doc]
print("SpaCy Tokens:", spacy_tokens)

# Compare SpaCy vs NLTK
print("\nComparison:")
print("NLTK handled abbreviations & punctuation well, but SpaCy handled email and URL tokens better:")
for token in spacy_tokens:
    print(token)
```

```python
print("\n----- TASK 6: Custom Tokenizer Function -----")

def custom_tokenizer(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text.split()

custom_tokens = custom_tokenizer(text)
print("Custom Tokens:", custom_tokens)
print("\nTotal Tokens:", len(custom_tokens))

print("\nComparison: Custom tokenizer is simple but removes punctuation and capitalization info.")
```

```python
print("\n----- TASK 7: Subword Tokenization (Advanced) -----")

# Using NLTK's Syllable Tokenizer approximation
words = ["processing", "tokenization", "university"]

def simple_syllable_split(word):
    return re.findall(r'[^aeiou]*[aeiou]+', word.lower())

for w in words:
    print(f"{w} -> {simple_syllable_split(w)}")
```

```python
print("\n----- TASK 8: Evaluation and Comparison -----")

from prettytable import PrettyTable

table = PrettyTable(["Method", "Handles Punctuation", "Handles Email", "Handles URL", "Preserves Case", "Ease of Use"])

table.add_row(["NLTK", "✔️ Contextual", "❌ Partial", "❌ Partial", "✔️", "Easy"])
table.add_row(["Regex", "❌ Removes", "❌ Breaks", "❌ Breaks", "❌", "Medium"])
table.add_row(["SpaCy", "✔️ Excellent", "✔️ Yes", "✔️ Yes", "✔️", "Medium-Easy"])
table.add_row(["Custom", "❌ Removes", "❌ Breaks", "❌ Breaks", "❌", "Very Easy"])

print(table)

print("\nConclusion:")
print("""SpaCy is the most accurate tokenizer because it understands abbreviations, emails, and URLs.
NLTK is simpler and effective for basic NLP tasks.
Regex and custom tokenizers are lightweight but lose linguistic information.
Trade-off: Accuracy (SpaCy) vs Simplicity (NLTK/Custom).""")
```
