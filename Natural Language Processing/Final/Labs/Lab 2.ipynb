```python
import re
import string
import nltk
from nltk.corpus import stopwords
```

```python
corpus = (
    "The University of Faisalabad is a private university located in Faisalabad, Pakistan. "
    "It was established in 2002 by the Madina Foundation. "
    "The university offers programs in engineering, health sciences, management, and social sciences. "
    "TUF is recognized by the Higher Education Commission of Pakistan. "
    "The university has modern laboratories, libraries, and research facilities. "
    "It focuses on producing skilled graduates to meet industry and societal needs. "
    "TUF places strong emphasis on research, innovation, and community service. "
    "The university encourages extracurricular activities, seminars, and workshops for student development. "
    "It has collaborations with national and international institutions. "
    "The University of Faisalabad continues to grow as a center of excellence in education and research."
)
```

```python
import nltk
nltk.download('stopwords')
```

```python
# Lower Corpus
corpus = corpus.lower()
# Remove Numbers
corpus = re.sub(r"\d+", "", corpus)
# Remove punctuation
translator = str.maketrans("", "", string.punctuation)
corpus = corpus.translate(translator)
# Remove stopwords
stop_words = set(stopwords.words("english"))
words = corpus.split()
filtered_words = [w for w in words if w not in stop_words]
```

```python
# Join back into a sentence
cleaned_text = " ".join(filtered_words)
```

```python
# Print Cleaned Data
print(cleaned_text)
```

```python
# 1. re.findall()
text1 = "Dog, Cat, Dog, Bird, Dog"
result1 = re.findall(r"Dog", text1)
print("findall:", result1)  
```

```python
# 2. re.compile()
pattern = re.compile(r"\d{2}/\d{2}/\d{4}")  # Date-like pattern
text2 = "The event is on 05/21/2025"
result2 = pattern.search(text2)
print("compile:", result2.group())  
```

```python
# 3. re.split()
text3 = "car;bus|train,plane"
result3 = re.split(r"[,;|]", text3)
print("split:", result3)  
```

```python
# 4. re.subn()
text4 = "one two one three one four"
result4 = re.subn(r"one", "zero", text4)
print("subn:", result4)  
```

```python
# 5. re.escape()
text5 = "Price is $5.99 (discount)"
escaped = re.escape(text5)
print("escape:", escaped)  
```

```python
# 6. re.search()
text6 = "Coding with Python is powerful!"
result6 = re.search(r"Python", text6)
print("search:", result6.group()) 
```
