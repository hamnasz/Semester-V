```python
print("\nEven Odd Function")
## Define Even Odd Function (Task 4)
def even_or_odd(x):
    if(x%2==0):
        print(x,"is Even Number")
    else:
        print(x,"is Odd Number")
number = int(input("Input Any Number: "))
even_or_odd(number)
```

```python
## Define Max Function (Task 5)
print("\nMin Max Function")
def max(x,y):
    if(x>y):
        print(x,"is Maximum")
    else:
        print(y,"is Maximum")
number1 = int(input("Input First Number: "))
number2 = int(input("Input Second Number: "))
max(number1,number2)
```
